```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Load the Iris dataset.
iris = datasets.load_iris()

print("Dataset shape:", iris.data.shape)
print("Feature names:", iris.feature_names)
print("Class names:", iris.target_names)
```

Select two features.

```python
# Keep every flower, but select only columns 0 and 2.
X_selected = iris.data[:, [0, 2]]

print("Selected features: sepal length and petal length")
print("Selected data shape:", X_selected.shape)
print("First five flowers:")
print(X_selected[:5])


```

Keep only two classes


```python
mask = iris.target != 2

X = X_selected[mask]
y = iris.target[mask]

print("Filtered measurements shape:", X.shape)
print("Filtered labels shape:", y.shape)
print("Remaining classes:", sorted(set(y.tolist())))
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
print("Train set shape:", X_train.shape)
print("Test set shape:", X_test.shape)
print("Train set labels:", y_train.shape)
print("Test set labels:", y_test.shape)

```

```python
model = LogisticRegression()

model.fit(X_train,y_train)

print("Model trained.")

```

Eval


```python
# Evaluate accuracy on the testing data.
accuracy = model.score(X_test, y_test)

# Get predictions so we can also count correct answers.
predictions = model.predict(X_test)
correct_predictions = (predictions == y_test).sum()

print("Accuracy:", accuracy)
print(f"Accuracy percentage: {accuracy:.2%}")
print("Correct predictions:", correct_predictions, "out of", len(y_test))
```

```python
feature_pairs = [
    (0, 1),
    (0, 2),
    (0, 3),
    (1, 2),
    (1, 3),
    (2, 3)
]

for pair in feature_pairs:
    # Select this pair of feature columns.
    X_pair = iris.data[:, list(pair)]

    # Keep the same two flower classes.
    X_pair = X_pair[mask]

    # Use the same split settings for every pair.
    X_train_pair, X_test_pair, y_train_pair, y_test_pair = train_test_split(
        X_pair,
        y,
        test_size=0.3,
        random_state=42
    )

    # Create and train a separate model for this pair.
    pair_model = LogisticRegression()
    pair_model.fit(X_train_pair, y_train_pair)

    # Evaluate this pair.
    pair_accuracy = pair_model.score(X_test_pair, y_test_pair)

    first_feature = iris.feature_names[pair[0]]
    second_feature = iris.feature_names[pair[1]]

    print(
        f"{first_feature} + {second_feature}: "
        f"{pair_accuracy:.2%}"
    )
```
